```python
a=input("Enter first element: ")
b=input("Enter second element: ")
```

```python
def sum(a, b):
    if((not a.isdecimal()) and a.replace('.','').isdecimal()):
        a=float(a)
    elif(a.isdecimal()):
        a=int(a)
    if((not b.isdecimal()) and b.replace('.','').isdecimal()):
        b=float(b)
    elif(b.isdecimal()):
        b=int(b)
    print(a+b)
```

```python
sum(a, b)
```
