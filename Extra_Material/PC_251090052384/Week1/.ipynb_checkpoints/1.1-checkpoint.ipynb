```python
a=int(input("Enter the first number: "))
b=int(input("Enter the second number: "))
```

```python
sum = a+b
prod = a*b
diff = a-b
div = a/b
idiv = a//b
rem = a%b
```

```python
print("Sum: ",sum)
print("Difference: ", diff)
print("Product: ", prod)
print("Division Quotient: ", idiv)
print("Division Remainder: ", rem)
print("Decimal Division: ", div)
```
