```python
s=input("Enter a string: ")
```

```python
ccount = 0
vcount = 0
for c in s:
    if(c.isalpha()):
        c=c.lower()
        if(c=='a' or c=='e' or c=='i' or c=='o' or c=='u'):
            vcount+=1
        else:
            ccount+=1
```

```python
print("Vowels: ", vcount)
print("Consonents: ", ccount)
```
