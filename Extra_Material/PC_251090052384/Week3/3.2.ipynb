```python
import numpy as np
```

```python
arr1 = np.random.randint(1,51,size=(3,3))
print(f"A:\n{arr1}")
arr2 = np.random.randint(1,51,size=(3,3))
print(f"\nB:\n{arr2}")
```

```python
print(f"A+B:\n{arr1+arr2}")
```

```python
print(f"A-B:\n{arr1-arr2}")
```

```python
print(f"Element-wise Multiplication:\n{arr1*arr2}")
```

```python
print(f"Dot Product:\n{np.dot(arr1,arr2)}")
```

```python
print(f"Transpose(A):\n{np.transpose(arr1)}")
print(f"Transpose(B):\n{np.transpose(arr2)}")
```

```python
det1 = round(np.linalg.det(arr1),10)
det2 = round(np.linalg.det(arr2),10)
print(f"Determinant(A): {det1}")
print(f"Determinant(B): {det2}")
```

```python
if(det1!=0):
    print(f"Inverse(A):\n{np.linalg.inv(arr1)}")
if(det2!=0):
    print(f"Inverse(A):\n{np.linalg.inv(arr2)}")
```
