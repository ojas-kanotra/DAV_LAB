```python
import numpy as np
```

```python
data = np.random.randint(1,51,size=(4,3))
print(data)
```

```python
print(f"mean: {data.mean()}")
print(f"median: {np.median(data)}")
print(f"standard deviation: {np.std(data)}")
```

```python
print(f"sum of all elements: {data.sum()}")
for i in range(data.shape[0]):
    print(f"sum or row {i+1}: {data[i].sum()}")
```

```python
print(f"reshaped matrix:\n{np.reshape(data, (6, 2))}")
```
