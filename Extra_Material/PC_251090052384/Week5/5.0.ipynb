```python
data = pd.DataFrame({
    'A': [1, -999, 3, -1000],
    'B': [-1000, 5, -999, 7]
})
print(data)
data.replace([-999.0,-1000], np.nan, inplace=True) 
print(data)
```

```python
data.fillna(data.median())
```

```python
data = pd.DataFrame({"food": ["salt", "bread", "butter",
                              "rice", "chips", "biscuits",
                              "chocolate", "honey", "sugar"],
                     "ounces": [4, 3, 12, 6, 7.5, 8, 3, 5, 6]})
data
```

```python
food_to_type = {
  "salt": "condiment",
  "bread": "breakfast",
  "butter": "condiment",
  "rice": "grocery",
  "chips": "snack",
  "honey": "condiment",
  "sugar": "condiment",
  "chocolate": "snack"
}
```

```python
data["ftype"] = data["food"].map(food_to_type)
data
```
