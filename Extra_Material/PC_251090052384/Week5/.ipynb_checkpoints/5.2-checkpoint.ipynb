```python
import numpy as np
import pandas as pd
```

```python
cerealdata = pd.DataFrame(pd.read_csv('cereal.csv'))
cerealdata
```

```python
cerealdata.describe()[3:]
```

```python
cerealdata.loc[:,"Protein (g)":"Vitamins and Minerals"]=cerealdata.loc[:,"Protein (g)":"Vitamins and Minerals"].replace(-1, cerealdata.mean(numeric_only=True))
```

```python
cerealdata.describe()[3:]
```

```python
cerealdata.loc[:,"Protein (g)":"Vitamins and Minerals"].fillna(cerealdata.median(numeric_only=True))
```
