```python
import numpy as np
import pandas as pd
```

```python
float_data = pd.Series([1.2, -3.5, np.nan, 0])
float_data
```

```python
string_data = pd.Series(["aaaaaaaaaaa", np.nan, None, "abcdefghij"])
string_data
```

```python
string_data.isna()
```

```python
string_data.notna()
```

```python
data1 = pd.Series([1, np.nan, 3.5, np.nan, 7])
data1
```

```python
data1.dropna()
```

```python
data2 = pd.DataFrame([[1., 6.5, 3.], [1., np.nan, np.nan],
                     [np.nan, np.nan, np.nan], [np.nan, 6.5, 3.]])
data2
```

```python
data2.dropna(axis="columns", how="all")
```

```python
df3 = pd.DataFrame(np.random.standard_normal((7, 3)))
df3.iloc[:4, 1] = np.nan
df3.iloc[:2, 2] = np.nan
df3.fillna(0)
```

```python
df4 = pd.DataFrame(np.random.standard_normal((7, 3)))
df4.iloc[:4, 1] = np.nan
df4.iloc[:2, 2] = np.nan
df4
```

```python
df4.fillna(df4.mean())
```

```python
data = pd.DataFrame({
    'A': [1, -999, 3, -1000],
    'B': [-1000, 5, -999, 7]
})
print(data)
data.replace([-999.0,-1000], np.nan, inplace=True) 
print(data)
```

```python
data.fillna(data.median())
```

```python
data5 = pd.DataFrame({"k1": ["one", "two"] * 3 + ["two"],
                     "k2": [1, 1, 2, 3, 3, 4, 4]})
data5
```

```python
data5.drop_duplicates()
```

```python
data5.drop_duplicates(subset=["k1"])
```
