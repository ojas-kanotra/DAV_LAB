```python
import numpy as np
import pandas as pd
```

```python
data = pd.DataFrame(pd.read_csv('Movies.csv'))
data
```

```python
ratings = data["Rating"]
print(data[data["Rating"]==ratings.max()])
```

```python
data[data["Language"]=="Hindi"].to_csv('HindiMovies.csv')
```
