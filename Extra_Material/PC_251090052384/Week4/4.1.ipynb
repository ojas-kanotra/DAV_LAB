```python
import numpy as np
import pandas as pd
```

```python
wtemps = pd.Series([25, 27, 24, 22, 25, 28, 27], index=['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday'])
```

```python
print(f'Average temperature: {wtemps.mean()}')
print(f'Maximum temperature: {wtemps.max()}')
print(f'Minimum temperature: {wtemps.min()}')
```

```python
print(wtemps[wtemps>wtemps.mean()])
```
