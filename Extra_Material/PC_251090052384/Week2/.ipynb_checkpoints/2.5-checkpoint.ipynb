```python
c1 = int(input('Enter no. of elements in first set: '))
s1=set()
print(f'Enter {c1} elements for first set:')
for i in range(c1):
    s1.add(input(f'{i+1}: '))
```

```python
c2 = int(input('Enter no. of elements in first set: '))
s2=set()
print(f'Enter {c2} elements for first set:')
for i in range(c2):
    s2.add(input(f'{i+1}: '))
```

```python
print(f'set1: {s1}')
print(f'set2: {s2}')
print("union:", s1 | s2)
print("intersection:", s1 & s2)
print("set1 - set2:", s1 - s2)
print("set2 - set1:", s2 - s1)
```
