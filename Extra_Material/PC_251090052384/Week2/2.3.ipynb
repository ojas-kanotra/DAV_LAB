```python
cricket = {1975: 'West Indies', 1979: 'West Indies', 1983: 'India', 1987: 'Australia', 1992: 'Pakistan', 1996: 'Sri Lanka', 1999: 'Australia', 2003: 'Australia', 2007: 'Australia', 2011: 'India', 2015: 'Australia', 2019: 'England', 2023: 'Australia'}
uniqueCountries = set(cricket.values())
wins = []
best = ''
m = 0
for c in uniqueCountries:
    w=list(cricket.values()).count(c)
    wins.append(w)
    if(m<w):
        best = c
        m=w
print('Wins\tCountry')
for i, c in enumerate(uniqueCountries):
    print(f'{wins[i]}\t{c}')
print(f'best country is {best} with {m} wins')
```
