```python
print("Enter numbers to be added into list. Enter 'q' to quit.")
l = []
while(True):
    i = input()
    if(i=='q'):
        break
    else:
        l.append(int(i))
print(l)
```

```python
print(f'max: {max(l)}')
print(f'min: {min(l)}')
avg = sum(i for i in l)/len(l)
print(f'average: {avg}')
```
