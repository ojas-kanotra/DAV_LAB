```python
# Name: Chhayank Sharma
# Registration no.: 60
# Write a Python code block that inputs a sentence from the user. 
# Count the frequency of each word in the sentence and store the result in a dictionary.
# Prints the dictionary with words as keys and their frequencies as values.
```

```python
S = input('Enter a sentence:\n')
```

```python
words = set(S.split())
```

```python
counts = []
for w in words:
    c=S.count(w)
    counts.append(c)
```

```python
dc=dict(zip(words, counts))
```

```python
print(dc)
```
