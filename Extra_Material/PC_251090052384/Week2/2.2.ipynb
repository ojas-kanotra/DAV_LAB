```python
uinp = input("Enter elements with spaces: ").split()
if(len(uinp)!=5):
    print("Error! Enter exactly 5 values")
    quit()
my_tuple = tuple(uinp)
print(my_tuple)
try:
    my_tuple[0]="Manipal"
except TypeError as e:
    print("Error Found! ", e)
    print("Cannot modify tuple, it is immutable")
```
