```python
dat_man= mtc[mtc["am"]==1]["mpg"] # Manual cars are marked as 1
print("Average MPG of manual cars:",dat_man.mean())
dat_auto= mtc[mtc["am"]==0]["mpg"]
print("Average MPG of automatic cars:",dat_auto.mean())
```

```python
mtc.boxplot(column='mpg', by = 'am')
plt.title('Fuel Efficiency Of Manual vs Automatic')
plt.xlabel('Transmission Type')
plt.ylabel('MPG')
plt.xticks([1, 2], ['Automatic', 'Manual'])
plt.show
```

```python
dat_x= mtc[["mpg","wt"]]
dat_x= dat_x.reset_index(drop=True)
dat_x.plot(kind='scatter', x='mpg', y='wt', color='blue', title='Weight vs MPG')
```
