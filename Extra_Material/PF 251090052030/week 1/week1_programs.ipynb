```python
"""1. Write a Python function to input two numbers and perform the 
Calculator operations of (+, -, *, /). """
def calc():
    a=int(input("Enter first number:"))
    b=int(input("enter second number:"))
    ch=input("""Choose which operation to perform:
                a for addition
                s for subraction
                m for multiplication
                d for division
                """)
    if ch=="a":
        print("addition=",a+b)
    elif ch=="s":
        print("subtraction=",a-b)
    elif ch=="m":
        print("multiplication=",a*b)
    elif ch=="d":
        print("Division=",a/b)
    else:
        print("Invalid option")
calc()
```

```python
"""2. Write a Python function that takes an integer and returns True if 
it’s a prime number and False otherwise"""
def prime(a):
    if a<=1:
        return False
    else:
        flag=False
        num=(int(a/2))+1
        for i in range(2,num):
            if a%i==0:
                flag=True
                break
            else:
                continue 
        return flag
a=int(input("Enter an integer:"))
prime(a)
```

```python
"""3. Create a Python function that creates a sequence between 1 and 
100 and prints all the odd numbers. Compute and display the sum 
of all the even numbers. """
def sequence():
    sum=0
    for i in range(1,101):
        if i%2:
            sum+=i
        else:
            print(i)
    print("sum of all even numbers:",sum)
sequence()
        
```

```python
"""4. Write a Python function to add two elements and display the 
result. The elements can be of type integer, float or string. """
def add():
    ch=input("""Enter type of data to add:
                s for string
                i for integer
                f for float:
                """)
    if ch=="s":
        a=input("Enter first string:")
        b=input("Enter second string:")
        return a+b
    elif ch=="i":
        a=int(input("Enter first integer:"))
        b=int(input("Enter second integer:"))
        return a+b
    elif ch=="f":
        a=float(input("Enter first integer:"))
        b=float(input("Enter second integer:"))
        return a+b
    else:
        return "Invalid choice"
add()
```

```python
"""5. Write a Python function that takes a string input from the user 
and counts the number of vowels and consonants in the string. """
def count():
    s=input("Enter a string:")
    l=['a','e','i','o','u','A','E','I','O','U']
    vc=0
    cc=0
    for i in s:
        if i in l:
            vc+=1
        else:
            cc+=1
    print("number of vowels are:",vc)
    print("number of consonants are:",cc)
count()
```

```python
"""write a python function that generates a list of numbers 1 to n and print all odd number or even number as choice based"""
def sequence():
    n=int(input("Enter the upper limit of list"))
    l = [i for i in range(1,n+1)]
    ch=input("""enter choice
                e to print all even numbers
                o to print all odd numbers
                """)
    if ch=="o":
        if n%2==0:
            for i in range(0,n,2):
                print(l[i])
        else:
            for i in range(0,n+1,2):
                print(l[i])
    else:
        for i in range(1,n,2):
                print(l[i])
       
sequence()
```

```python
#write a python function  that takes an integer and display sum of digits
def sum(a):
    num=a
    sum=0
    while num!=0:
            sum+=(num%10)
            num=num//10
    return sum
a=int(input("Enter an integer:"))
sum(a)
```

```python
#write a python function that takes a number and return the reverse of it
def reverse():
    a=int(input("enter an integer:"))
    num=a
    rev=""
    while num!=0:
        rev+=str(num%10)
        num//=10
    return int(rev)
reverse()
```
