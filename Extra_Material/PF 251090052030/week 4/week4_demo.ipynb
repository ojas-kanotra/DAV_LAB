```python
values=[10,20,30,40]
labels=["A","B","C","D"]
S=pd.Series(values,index=labels)
S
```

```python
## INSPECTING A SERIES
print("index:",S.index)
print("Values:",S.values)
```

```python
## aggregation functions
## pandas provide built in function to summarize data
print("mean:",S.mean())
print("max:",S.max())
print("min:",S.min())
```

```python
S.idxmin()
```
