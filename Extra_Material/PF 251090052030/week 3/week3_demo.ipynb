```python
x=np.array([[2,8,1],
           [5,3,9]])
print("x:\n",x)
print("\nmean:",np.mean(x))
print("Median:",np.median(x))
print("Std:",np.std(x))
```

```python
#create a random 3x4 matrix and compute mean,median and standard deviation
arr=np.random.randint(1,10,size=(3,4))
print("array:\n",arr)
print("\nmean:",np.mean(arr))
print("Median:",np.median(arr))
print("Std:",np.std(arr))
```

```python
#axis=0 means down the rows
#axis=1 means across the column
print("x:\n",x)
print("\nsum of all elements:",np.sum(x))
print("row wise sum:",np.sum(x,axis=1))
print("column wise sum:",np.sum(x,axis=0))
```

```python
flat=np.arange(1,13)
M=flat.reshape(3,4)
print("flat:",flat)
print("\nreshaped to 3x4:\n",M)
print("reshaped to 2x6:\n",M.reshape(2,6))
```

```python
z=np.array([[1,2,3],
            [4,5,6]])
add_row=np.array([10,20,30])
add_col=np.array([[100],[200]])
print("Z:\n",z)
print("\nZ + add row:\n",z+add_row)
print("\nZ + add column:\n",z+add_col)
```

```python
M1=np.array([[1,2,3],
            [8,1,4],
            [5,6,9]])

M2=np.array([[5,3,7],
            [2,9,2],
            [6,1,8]])
print("add M1 and M2:",M1+M2)
print("subtract M1 and M2:",M1-M2)
```

```python
import numpy as np
A=([[1,1],
     [1,2]])
det_A=np.linalg.det(A)
print("determinant of A:",det_A)
inv_A=np.linalg.inv(A)
print("inverse of A:",inv_A)
```
