```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("mtcars.csv")
print(df.shape)
print(df.info())
print(df.describe)
```

```python
print("total cars",len(df))
print("unique cylinder types:",df["cyl"].nunique())
print("Transmission Types:",df["am"].nunique())
```

```python
cyl_counts=df.groupby("cyl")["mpg"].count()
cyl_counts
```

```python
mpg=df["mpg"]
print(mpg.plot())
```

```python
df=df[['cyl','wt','mpg']]
df.plot()
```

```python
cyl_counts=df['cyl'].value_counts()
cyl_counts.plot(kind="bar",color="silver")
plt.title("Count of cars by cylinder")
plt.xlabel("cylinders")
plt.ylabel("count")
plt.show()
```

```python
df.plot(
kind="scatter",
x="wt",
y="mpg",
figsize=(9,9),
color="pink")
```

```python
plt.figure(figsize=(8,8))
plt.boxplot(x="cyl",y="mpg",data=df)
plt.title("mpg by cylinder category ")
plt.xlabel("count of cylinders")
plt.ylabel("MPG")
```
