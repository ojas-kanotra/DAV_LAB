```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("mtcars.csv")
print(df.shape)
print(df.info())
print(df.describe)
```

```python
print("total cars",len(df))
print("unique cylinder types:",df["cyl"].nunique())
print("Transmission Types:",df["am"].nunique())
```

```python
cyl_counts=df.groupby("cyl")["mpg"].count()
cyl_counts
```

```python
mpg=df["mpg"]
print(mpg.plot())
```

```python
df=df[['cyl','wt','mpg']]
df.plot()
```

```python
cyl_counts=df['cyl'].value_counts()
cyl_counts.plot(kind="bar",color="silver")
plt.title("Count of cars by cylinder")
plt.xlabel("cylinders")
plt.ylabel("count")
plt.show()
```

```python
df.plot(
kind="scatter",
x="wt",
y="mpg",
figsize=(9,9),
color="pink")
```

```python
plt.figure(figsize=(8,8))
sns.boxplot(x="cyl",y="mpg",data=df)
plt.title("mpg by cylinder category ")
plt.xlabel("count of cylinders")
plt.ylabel("MPG")
```

```python
cyl_counts=df['cyl'].value_counts()
plt.figure(figsize=(8,8))
plt.pie(
    cyl_counts,
    labels=cyl_counts.index,
    autopct='%1.1f%%',
    colors=["skyblue","lightyellow","pink"],
    startangle=140,
    explode=[0.1,0,0],
)
plt.title("proportion of cars by number of cylinders")
plt.show()
```

```python
x = df['mpg']
# Plotting the histogram
plt.hist(x)
# Adding title to the plot
plt.title("Distribution of MPG in Cars Dataset")
# Adding label on the y-axis (Frequency of each cylinder count)
plt.ylabel('Frequency')
plt.xlabel('MPG')
# Adding label on the x-axis (Number of cylinders in cars)
# Displaying the plot
plt.show()
```

```python
 # Pie Chart for Transmission Types
# Reload the original dataframe to include the 'am' column
df = pd.read_csv('mtcars.csv')
am_counts = df['am'].value_counts()
# Create a pie chart
plt.figure(figsize=(8, 8))
plt.pie(
am_counts,
labels=['Automatic', 'Manual'], # Custom labels
autopct='%1.1f%%',
colors=['brown', 'silver'],
startangle=90
)
plt.title('Proportion of Cars by Transmission Type')
plt.show()
```

```python
# Nested Pie Chart which needs Data preparation
outer_labels = df['cyl'].value_counts().index
outer_sizes = df['cyl'].value_counts().values
inner_labels = df['gear'].value_counts().index
inner_sizes = df['gear'].value_counts().values
# Create nested pie chart
fig, ax = plt.subplots(figsize=(8, 8))
ax.pie(outer_sizes, labels=outer_labels, radius=1, autopct='%1.1f%%',colors=['skyblue', 'lightgreen', 'pink'])
ax.pie(inner_sizes, labels=inner_labels, radius=0.7, autopct='%1.1f%%',colors=['silver', 'lightbrown', 'golden'])
ax.set(aspect="equal")
plt.title('Nested Pie Chart: Cylinders (Outer) and Gears (Inner)')
plt.show()
```
