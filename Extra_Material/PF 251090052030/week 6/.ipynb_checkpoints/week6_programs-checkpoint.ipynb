```python
"""1. For the MTCARS dataset, answer the specified questions with 
summarization and effective visuals. """
#. Load the MTCARS dataset and display its basic structure (rows, columns, data types).
import pandas as pd
dat_mtcars=pd.read_csv("mtcars.csv", index_col=0)
dat_mtcars
```

```python
dat_mtcars.info()
```

```python
""" Compute summary statistics (mean, median, min, max) for:
• Miles per gallon (mpg)
• Horsepower (hp)
• Weight (wt)"""
print(dat_mtcars[['mpg', 'hp', 'wt']].describe())
print("\nmedian of mpg,hp,wt:")
print(dat_mtcars[['mpg', 'hp', 'wt']].median())
```

```python
# 1. Explore the distribution of Fuel efficiency of the cars
dat_mtcars["mpg"].plot(kind='hist', bins=10, title="Distribution of fuel efficiency of cars")
```

```python
dat_mtcars["mpg"].plot(kind='box')
```

```python
# 2. Find the cars with the best and worst fuel efficiency
max_mpg=dat_mtcars["mpg"].max()
print("cars with best fuel efficiency:")
dat_mtcars[dat_mtcars["mpg"]==max_mpg]
```

```python
min_mpg=dat_mtcars["mpg"].min()
print("cars with worst fuel efficiency:")
dat_mtcars[dat_mtcars["mpg"]==min_mpg]
```

```python
# 3. How many cars are manual vs automatic?
dat_mtcars["am"].nunique()
```

```python
print("maual:0 vs automatic:1")
dat_mtcars["am"].value_counts()
```

```python
# 4. Find the car with the worst horsepower
min_hp=dat_mtcars["hp"].min()
print("cars with worst horse power:")
dat_mtcars[dat_mtcars["hp"]==min_hp]
```

```python
# 5. Find 5 number summary and draw the boxplot of displacement.
dat_mtcars["disp"].plot(kind='box')
```

```python
num_col=["mpg","cyl","disp","hp","drat","wt","qsec","vs","am","gear","carb"]
dat_mtcars[num_col].quantile([0,0.25,0.5,0.75,1.0])
```

```python
#6. Which is the heaviest car? How many gears does it have?
max_wt = dat_mtcars["wt"].max()
heaviest_car = dat_mtcars[dat_mtcars["wt"] == max_wt]
print("Heaviest Car:")
print(heaviest_car[["wt", "gear"]])
```

```python
# 7. Car with best qsec (fastest acceleration)
max_qsec = dat_mtcars["qsec"].max()
print("Car with Best QSEC:")
print(dat_mtcars[dat_mtcars["qsec"] == max_qsec])
```

```python
# 8. Average MPG for manual vs automatic
dat_man = dat_mtcars[dat_mtcars["am"] == 1]["mpg"]
dat_auto = dat_mtcars[dat_mtcars["am"] == 0]["mpg"]

print("Average MPG of Manual Cars:", dat_man.mean())
print("Average MPG of Automatic Cars:", dat_auto.mean())
```

```python
# 9. Side-by-side boxplot (Manual vs Automatic)
import matplotlib.pyplot as plt
dat_mtcars.boxplot(column="mpg", by="am")
plt.title("MPG by Transmission Type")
plt.suptitle("")
plt.xlabel("Transmission (0 = Auto, 1 = Manual)")
plt.ylabel("MPG")
plt.show()
```

```python
# 10. Relationship between Weight and MPG (Scatter)
import matplotlib.pyplot as plt
dat_mtcars.plot(
    kind='scatter',
    x='mpg',
    y='wt',
    title='Weight vs MPG'
)
plt.xlabel("Miles per Gallon (MPG)")
plt.ylabel("Weight (1000 lbs)")
plt.show()
```

```python
# 11. Categorize cars based on number of gears
print("Cars by Gear Type:")
print(dat_mtcars["gear"].value_counts())

```

```python
# 12. Relationship between MPG and Gears
import matplotlib.pyplot as plt
dat_mtcars.boxplot(column="mpg", by="gear")
plt.title("MPG by Number of Gears")
plt.suptitle("")
plt.xlabel("Number of Gears")
plt.ylabel("MPG")
plt.show()
```

```python
# 13. Relationship between Horsepower and Cylinders
import matplotlib.pyplot as plt
dat_mtcars.plot(
    kind='scatter',
    x='cyl',
    y='hp',
    title='Horsepower vs Cylinders'
)
plt.xlabel("Number of Cylinders")
plt.ylabel("Horsepower")
plt.show()
```

```python
# 14. Relationship between Displacement and Horsepower
import matplotlib.pyplot as plt
dat_mtcars.plot(
    kind='scatter',
    x='disp',
    y='hp',
    title='Displacement vs Horsepower'
)
plt.xlabel("Displacement")
plt.ylabel("Horsepower")
plt.show()
```

```python

# 15. Best balance of high speed (hp) and good fuel efficiency (mpg)

# Create a simple score combining normalized hp and mpg
normalized = (dat_mtcars[["hp", "mpg"]] - dat_mtcars[["hp", "mpg"]].mean()) / dat_mtcars[["hp", "mpg"]].std()
dat_mtcars["balance_score"] = normalized["hp"] + normalized["mpg"]

best_choice = dat_mtcars.sort_values("balance_score", ascending=False).head(1)

print("Best Car for High Speed and Good Fuel Efficiency:")
print(best_choice[["hp", "mpg", "balance_score"]])
```
