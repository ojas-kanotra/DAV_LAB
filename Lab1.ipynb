### 1. Write a Python function to input two numbers and perform the Calculator operations of (+, -, *, /). 

```python
def calculator(a,b,operation):
    print(f"{a} {operation} {b} =", eval(f"{a}{operation}{b}"))
calculator(int(input("Enter the first number: ")), int(input("Enter the second number: ")), input("Enter the operator: "))
```

### 2. Write a Python function that takes an integer and returns True if it’s a prime number and False otherwise. 

```python
def prime_checker(a):
    for i in range(2, a):
        if a%i==0:
            return False
    return True

print(prime_checker(int(input("Enter the number to be checked: "))))
```

### 3. Create a Python function that creates a sequence between 1 and 100 and prints all the odd numbers. Compute and display the sum of all the even numbers. 

```python
def sequence():
    sum=0
    for i in range(1, 100):
        if i%2==0:
            sum+=i
        else:
            print(i)
    print("Sum of numbers is: ", sum)
sequence()
```

### 5. Write a Python function that takes a string input from the user and counts the number of vowels and consonants in the string

```python
def counter(str):
    consonant=0
    vovel=0
    for i in str:
        if i in ['a', 'e', 'i', 'o', 'u']:
            vovel+=1
        elif(i==' ' or i=='!' or i=='.'):
            continue
        else:
            consonant+=1
    print("Total consonant count is: ", consonant)
    print("Total vovel count is: ", vovel)
    
counter(input("Enter the string: "))
```
