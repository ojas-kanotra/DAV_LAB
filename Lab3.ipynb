### Q1. Generate a 3x4 NumPy array with random integers between 1 and 50. 

```python
arr = np.random.randint(1, 50, size=(3,4))
print(arr)
```

> A. Calculate and print the Mean, Median, and Standard Deviation 
of the array 

```python
print("Mean is : ", np.mean(arr))
print("Median is: ", np.median(arr))
print("Standard deviation is: ", np.std(arr))
```

> B. Print the Sum of all elements and the sum of each row. 

```python
print("Sum of elements is: ", np.sum(arr))
print("Sum of row 1: ", np.sum(arr[0]))
print("Sum of row 1: ", np.sum(arr[1]))
print("Sum of row 1: ", np.sum(arr[2]))
```

> C. Reshape the 3x4 array into a 2x6 array and print it. 

```python
arr.reshape(2, 6)
print(arr)
```

### Q2. Create two (3 * 3) matrices using NumPy and print it. Perform and print the results of the following linear algebra operations 

```python
arr1 = np.random.randint(100, size=(3,3))
arr2 = np.random.randint(100, size=(3,3))
print("Matrix 1: \n", arr1)
print("Matrix 2: \n", arr2)
```

> A. Matrix addition 

```python
print("Addition of matrix 1 and matrix 2 is: \n", arr1+arr2)
```

> B. Matrix subtraction 

```python
print("Subtraction of matrix 1 and matrix 2 is: \n", arr1-arr2)
```

> C. Matrix multiplication (element-wise and dot product) 

```python
print("Multiplication of matrix 1 and matrix 2 is: \n", arr1*arr2)
```

> D. Transpose of a matrix 

```python
print("Transpose of matrix 1 is: \n", np.transpose(arr1))
print("Transpose of matrix 2 is: \n", np.transpose(arr2))
```

> E. Determinant and inverse (if applicable) 

```python
det1 = np.linalg.det(arr2)
det2 = np.linalg.det(arr1)
print("Determinant of matrix 1 is: \n", det1)
print("Determinant of matrix 2 is: \n", det2)
if det1!=0 and det2!=0:
    print("Inverse of matrix 1 is: \n", np.linalg.inv(arr1))
    print("Inverse of matrix 2 is: \n", np.linalg.inv(arr2))
```
