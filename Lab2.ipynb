### 1. Write a Python code block that inputs numbers into a list. Print the largest, smallest, the sum, and the average of the numbers. Count occurrences of a specific number in the list.

```python
n = int(input("Enter the number of elements: "))
data = []
for i in range(n):
    data.append(int(input(f"Sr No. {i}: ")))
print("Largest number is: ", max(data))
print("Smallest number is: ", min(data))
print("Sum of numbers is: ", sum(data))
print("Averga of the numbers is: ", sum(data)/len(data))
```

### 3. Write a Python code block to create a dictionary of cricket World Cup winners. Let the key be the year; the value is the country that won the World Cup that year. Print the name of the best- performing country. Display the unique list of countries that have won the World Cup. 

```python
world_cup_winners = {
    1975: "West Indies",
    1979: "West Indies",
    1983: "India",
    1987: "Australia",
    1992: "Pakistan",
    1996: "Sri Lanka",
    1999: "Australia",
    2003: "Australia",
    2007: "Australia",
    2011: "India",
    2015: "Australia",
    2019: "England",
    2023: "Australia"
}
val = list(world_cup_winners.values())
uni_val = list(set(world_cup_winners.values()))
counts = list(map(val.count, uni_val))    
best_performer = uni_val[counts.index(max(counts))]
print("Unique list of winners:", uni_val)
print("Name of best performing country is:", best_performer)
```

### 4. Write a Python code block that inputs a sentence from the user. Count the frequency of each word in the sentence and store the result in a dictionary. Prints the dictionary with words as keys and their frequencies as values. 

```python
sentence = input("Enter the sentence: ")
sentence = sentence.split(" ")
uni_val = list(set(sentence))
counts = list(map(sentence.count, uni_val))
dic = {}
for i in range(len(sentence)):
    if sentence[i] not in dic.values():
        dic[sentence[i]] = sentence.count(sentence[i])
print(dic)
```

### 5. Write a Python code block to input numbers into two sets. Perform union, intersection, and difference operations on the sets and print the results. 

```python
m = int(input("Enter the number of elements in first set: "))
n = int(input("Enter the number of elements in second set: "))
set_1 = set()
set_2 = set()
for i in range(m):
    set_1.add(input(f"Enter element {i} of first set: "))
for i in range(n):
    set_2.add(input(f"Enter element {i} of second set: "))
print("Union of two sets is:", set_1.union(set_2))
print("Intersectiopn of two sets is:", set_1.intersection(set_2))
print("Difference of two sets is:", set_1.difference(set_2))
```
