#  Demo 7: Plotting using Matplotlib & Seaborn 

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

Q1. Load the IPL dataset and display first 10 records.  


```python
data = pd.read_csv('data_CC.csv')
df = pd.DataFrame(data)
data.head(10)
```

Q2. Find the total number of:  
- Matches
- Teams
- Seasons

```python
print("Total Matches conducted are: ", np.shape(data)[0])
print("Total number of teams are: ", len(set(df['team1'])|set(df['team2'])))
print("Total number of seasons are: ", len(df['season'].unique()))
```

Q3. Plot the number of matches played per season using Matplotlib. 

```python
df['season'].value_counts().sort_index().plot(kind='bar')
plt.xlabel("Season")
plt.ylabel("Number of Matches")
plt.title("Matches per Season")
plt.show()
```

Q4. Use Seaborn to visualize:
> Top 5 teams with most wins 

```python
top_teams = df['winner'].value_counts().head(5)

sns.barplot(x=top_teams.values, y=top_teams.index)

plt.xlabel("Number of Wins")
plt.ylabel("Team")
plt.title("Top 5 Teams with Most Wins")

plt.show()
```

Q6. Plot a pie chart representing match wins by venue. 



```python
vc = df['venue'].value_counts().head(10)
plt.pie(vc.values, labels=vc.index,  autopct='%1.1f%%')
plt.title("Simple Pie Chart")
plt.show()
```

Q7. Compare home vs away wins using an appropriate plot. 

```python
df['home_win'] = df['winner'] == df['team1']
df['away_win'] = df['winner'] == df['team2']

home_wins = df['home_win'].sum()
away_wins = df['away_win'].sum()

sns.barplot(x=["Home_Win", "Away_Win"], y=[home_wins, away_wins])

plt.title("Home vs Away Wins")
plt.show()
```
